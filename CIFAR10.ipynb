```python
import tensorflow as tf
import numpy as np
```

```python
tf.keras.backend.clear_session()
```

```python
model = tf.keras.Sequential()
```

```python
model.add(tf.keras.layers.Input(shape=[32,32,3]))
model.add(tf.keras.layers.Flatten())
model.add(tf.keras.layers.BatchNormalization())
```

```python
for _ in range(20):
  model.add(tf.keras.layers.Dense(100, kernel_initializer="he_normal"))
  model.add(tf.keras.layers.BatchNormalization())
  model.add(tf.keras.layers.Activation("swish"))
```

```python
loaded_data = tf.keras.datasets.cifar10.load_data()
```

```python
(X_train_full, y_train_full), (X_test, y_test) = loaded_data
```

```python

tf.random.set_seed(42)
```

```python
X_train_full.shape
```

```python
model.add(tf.keras.layers.Dense(10, activation="softmax"))
```

```python
model.summary()
```

```python
optimizer = tf.keras.optimizers.Nadam(learning_rate=5e-5)
model.compile(loss="sparse_categorical_crossentropy",
              optimizer=optimizer,
              metrics=["accuracy"])
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python

early_stopping_criteria = EarlyStopping(patience=20, restore_best_weights=True)
```

```python
# validation_split=0.1,
history = model.fit(X_train_full, y_train_full, validation_split=0.1, callbacks=[early_stopping_criteria], epochs=100)
```

```python
# 37 epochs to reach 0,5423
# mit Batch Normalisierung direkt 32 epochs to reach 0,7184

print("Best val_accuracy:", max(history.history["val_accuracy"]))
```
