```python
# Converting Boolean Values to numbers. Much easier to handle
X_train["VIP"] = X_train["VIP"].astype('float')
X_train["CryoSleep"] = X_train["CryoSleep"].astype('float')
X_test["VIP"] = X_train["VIP"].astype('float')
X_test["CryoSleep"] = X_train["CryoSleep"].astype('float')
```

```python
# Imputing with scikit learn. Also orders numerical values and categorical values
from sklearn.impute import SimpleImputer
from sklearn.pipeline import Pipeline
from sklearn.compose import ColumnTransformer
```

```python
num_columns_mean = ["Age", "RoomService", "FoodCourt", "ShoppingMall", "Spa", "VRDeck"]
num_columns_most_frequent = ["VIP", "CryoSleep"]
cat_columns = ["HomePlanet", "Destination"]
```

```python
cat_pipeline = Pipeline([
    ("categorical_mf_imputer", SimpleImputer(strategy="most_frequent"))
])

num_mean_pipeline = Pipeline([
    ("numerical_mean_imputer", SimpleImputer(strategy="mean"))
])

num_mf_pipeline = Pipeline([
    ("numerical_mf_imputer", SimpleImputer(strategy="most_frequent"))
])
```

```python
column_trans = ColumnTransformer([
    ("num_mean_transformer", num_mean_pipeline, num_columns_mean),
    ("num_mf_transformer", num_mf_pipeline, num_columns_most_frequent),
    ("cat_transformer", cat_pipeline, cat_columns)
])
```

```python
X_train_imputed = column_trans.fit_transform(X_train) # Convertes into a numpy array
```

```python
X_test_imputed = column_trans.fit_transform(X_test)
```

These input values can now be processed with a keras model. Imputing, droping and separating must still be processed with new data

```python
num_inputs_train = X_train_imputed[:,:8]
planet_inputs_train = X_train_imputed[:,8:9]
destination_inputs_train = X_train_imputed[:,9:]
num_inputs_test = X_test_imputed[:,:8]
planet_inputs_test = X_test_imputed[:,8:9]
destination_inputs_test = X_train_imputed[:,9:]
```

```python
# Importing Keras and Tensorflow libraries
import tensorflow as tf

tf.random.set_seed(42)

tf.keras.backend.clear_session()
```

```python
# Defining a layer to string encode and embedd string lookups

# Every Categorical Feature need a lookup layer that is adapted to the possible data
home_planet_lookup_layer = tf.keras.layers.StringLookup()
destination_lookup_layer = tf.keras.layers.StringLookup()

# Adapting
home_planet_lookup_layer.adapt(planet_inputs_train)
destination_lookup_layer.adapt(destination_inputs_train)

# Output number of unique cats/2 and perform math.ceil()
lookup_and_embed_home_planet = tf.keras.Sequential([
    home_planet_lookup_layer,
    tf.keras.layers.Embedding(input_dim=home_planet_lookup_layer.vocabulary_size(), output_dim=2),
    tf.keras.layers.Flatten()
])

lookup_and_embed_destination = tf.keras.Sequential([
    destination_lookup_layer,
    tf.keras.layers.Embedding(input_dim=destination_lookup_layer.vocabulary_size(), output_dim=2),
    tf.keras.layers.Flatten()
])
```

```python
# Adapting the numerical layer

numerical_features_norm_layer = tf.keras.layers.Normalization()
numerical_features_norm_layer.adapt(num_inputs_train)
```

```python
# defining a functional model

# input layers
# (samples, number of features) --> Keras handles samples automatically through batch_size in model.fit()
# it might be confusing that numpy indexing and shape information is vice versa. The first digit is the index of the record (sample) and the second one the index of the vector element
# but in keras the first element is the number of features in each vector
num_input_layer = tf.keras.layers.Input(shape=(8,), name="num")
homeplanet_input_layer = tf.keras.layers.Input(shape=(1,), dtype=tf.string, name="homeplanet")
destination_input_layer = tf.keras.layers.Input(shape=(1,), dtype=tf.string, name="destination")

# embeddings of categorical data
embedded_home_planet = lookup_and_embed_home_planet(homeplanet_input_layer)
embedded_destination = lookup_and_embed_destination(destination_input_layer)

# normalization of numerical data
normalized_numerical_inputs = numerical_features_norm_layer(num_input_layer)

# concat normalized and embedded values
x = tf.keras.layers.concatenate([normalized_numerical_inputs, embedded_home_planet, embedded_destination])

x = tf.keras.layers.Dense(50, activation="swish", kernel_initializer="he_normal")(x)
x = tf.keras.layers.BatchNormalization()(x)

x = tf.keras.layers.Dense(50, activation="swish", kernel_initializer="he_normal")(x)
x = tf.keras.layers.BatchNormalization()(x)

x = tf.keras.layers.Dense(25, activation="swish", kernel_initializer="he_normal")(x)
x = tf.keras.layers.BatchNormalization()(x)

output = tf.keras.layers.Dense(1, activation="sigmoid", name="output")(x)

model = tf.keras.models.Model(inputs=[num_input_layer, homeplanet_input_layer, destination_input_layer], outputs=[output])
model.compile(loss="binary_crossentropy", metrics=["accuracy"], optimizer="adam")
```

```python
model.summary()
```

```python
from tensorflow.keras.callbacks import EarlyStopping
early_stopping_callback = EarlyStopping(monitor="val_loss", patience=10, restore_best_weights=True)
```

```python
# !!!The String Lookup Layer needs tensorflow strings. Numpy strings are not compatible. We need to convert each categorical string input to a tensorflow constant
history = model.fit(
    (
        num_inputs_train.astype('float32'),
        tf.constant(planet_inputs_train, dtype=tf.string),
        tf.constant(destination_inputs_train, dtype=tf.string),
    )
    , y_train.astype('float32')         # the boolean True False Output is also converted to a numerical Value. If there werde
    , epochs=128
    , batch_size=32
    , validation_split=0.1
    , callbacks=[early_stopping_callback]
)
```

```python
# Evaluation --> We do not have a test classification result set
```

```python
# Prediction
# using the same input like in fit method is mandatory
# We need to pass the data in the same order, that is present in the model constructor: model = tf.keras.models.Model(inputs=[num_input_layer, homeplanet_input_layer, destination_input_layer], outputs=[output]) and the fit call
y_result = model.predict((
    num_inputs_test.astype('float32'),
    tf.constant(planet_inputs_test, dtype=tf.string),
    tf.constant(destination_inputs_test, dtype=tf.string),
))
```
