```python
import numpy as np
```

```python
from sklearn.tree import DecisionTreeRegressor
```

```python
np.random.seed(42)
randArray = np.random.rand(200, 1)
```

```python
print(randArray.min())
print(randArray.max())
```

```python
X_Quad = randArray - 0.5
```

```python
print(X_Quad.min())
print(X_Quad.max())
```

```python
y_quad = X_Quad ** 2 + 0.025 * np.random.randn(200,1)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(6, 4))
plt.scatter(X_Quad, y_quad)
plt.xlabel('x')
plt.ylabel('y')
plt.title('Quadratic data (points only)')
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
import seaborn as sns
```

```python
y_quad[:,0]
```

```python
sns.scatterplot(x=X_Quad[:, 0], y=y_quad[:, 0])
```

```python
tree_reg = DecisionTreeRegressor(max_depth=2, random_state=42)
```

```python
tree_reg.fit(X_Quad, y_quad)
```

```python
from sklearn.tree import export_graphviz

export_graphviz(tree_reg, out_file='regtree_visual.dot', feature_names=["X-Val"], rounded=True, filled=True, class_names=["Y-Val"])
# File can be checked with online viewer
```
