```python
import tensorflow as tf
import numpy as np
```

```python
tf.keras.backend.clear_session()
```

```python
model = tf.keras.Sequential()
```

```python
model.add(tf.keras.layers.Input(shape=[32,32,3]))
model.add(tf.keras.layers.Flatten())
model.add(tf.keras.layers.BatchNormalization())
```

```python
for _ in range(5):
  model.add(tf.keras.layers.Dropout(rate=0.1))
  model.add(tf.keras.layers.Dense(32, kernel_initializer="he_normal"))
  model.add(tf.keras.layers.BatchNormalization())
  model.add(tf.keras.layers.Activation("swish"))
```

```python
loaded_data = tf.keras.datasets.cifar10.load_data()
```

```python
(X_train_full, y_train_full), (X_test, y_test) = loaded_data
```

```python

tf.random.set_seed(42)
```

```python
X_train_full.shape
```

```python
model.add(tf.keras.layers.Dropout(rate=0.2))
model.add(tf.keras.layers.Dense(10, activation="softmax"))
```

```python
model.summary()
```

```python
# _____ Learning Rate Sheduling _____

# Exponential learning rate sheduling
# Computed as learning rate * decay_rate ** (epoch/decay_steps)
# Add to opimizer
# initial_learning_rate = 1e-3
# lr_schedule = tf.keras.optimizers.schedules.ExponentialDecay(
#     initial_learning_rate,
#     decay_steps=5,
#     decay_rate=0.1,
#     staircase=True)

# Performance Sheduling
# Add to model fit callbacks
reduce_lr = tf.keras.callbacks.ReduceLROnPlateau(monitor='val_loss', factor=0.2, patience=5)


```

```python
#optimizer = tf.keras.optimizers.AdamW(learning_rate=1e-2, weight_decay=0.01)
optimizer = tf.keras.optimizers.Adam(learning_rate=1e-2)
model.compile(loss="sparse_categorical_crossentropy", optimizer=optimizer, metrics=["accuracy"])
```

```python
from tensorflow.keras.callbacks import EarlyStopping
```

```python

early_stopping_criteria = EarlyStopping(patience=10, restore_best_weights=True)
```

```python
# validation_split=0.1,
history = model.fit(X_train_full, y_train_full, validation_split=0.1, callbacks=[early_stopping_criteria, reduce_lr], batch_size=32, epochs=100)
```

```python
# Nut mit Batch Normalisierung: Best val_accuracy: 0.39239999651908875
# Mit Drop out 0.10019999742507935 --> Ganz mies
# mit Exponential Learning Rate --> Best val_accuracy: 0.11379999667406082
# AdamW und Performance Sheduling --> Best val_accuracy: 0.49480000138282776
# Adam mit Performance Sheduling --> Best val_accuracy: 0.4912000000476837
# Adam mit Performance Sheduling und Dropout --> Best val_accuracy: 0.5156000256538391

print("Best val_accuracy:", max(history.history["val_accuracy"]))
```

```python
model.evaluate(X_test, y_test)
```
